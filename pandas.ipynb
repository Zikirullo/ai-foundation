```python
data_with_missings = {
    "cars": ["BMW", "Volvo", None, "Tesla", "Ferari", None],
    "counts": [3, 7, None, 8, 9, None]
}
df_missing = pd.DataFrame(data_with_missings)
print(df_missing)
```

```python
# isnull() + sum() = how many raws have null value

df_missing.isnull().sum()
```

```python
# dropna() drops rows with no data

df_with_data = df_missing.dropna()

print(df_with_data)
```

```python
df_missing.dropna(how="all")
df_missing.dropna(subset="counts")
```

```python
# fillna() -> fills row with no data

df_missing.fillna(0)
df_missing.fillna({"cars": "unknown", "counts": 0})

```
